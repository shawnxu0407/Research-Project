```python
import sympy as sp
from sympy import expand
```

```python
A, B, C, D, E, F, G = sp.symbols('A, B, C, D, E, F, G')
x = sp.symbols('x')
k = sp.symbols('k')
d_1=sp.symbols('d_1')
d_2=sp.symbols('d_2')
d_3=sp.symbols('d_3')
d_4=sp.symbols('d_4')
```

```python
expand(((C-A)**3)*(B**4)*(A**4)-(A+1)**4)
```

## 2 Species System

```python
# Define symbolic matrix A
ODE_Jacobian_2D = sp.Matrix([
    [-A-x, -B],
    [A, B-x]
])

PDE_Jacobian_2D = sp.Matrix([
    [-A-d_1*(k**2)-x, -B],
    [A, B-d_2*(k**2)-x]
])
```

```python
ODE_Jacobian_2D.det()
```

```python
PDE_Jacobian_2D.det()
```

## 3 Species System

```python
# Define symbolic matrix A
ODE_Jacobian_3D = sp.Matrix([
    [-A-x, -B+C, E],
    [A, B-C-D-x, -F],
    [0, D, 0-x]
])

PDE_Jacobian_3D = sp.Matrix([
    [-A-d_1*(k**2)-x, -B+C, E],
    [A, B-C-D-d_2*(k**2)-x, -F],
    [0, D, 0-d_3*(k**2)-x]
])
```

```python
det=ODE_Jacobian_3D.det()


det_expanded = sp.expand(det)
det_collected = sp.collect(det_expanded, x)
det_collected
```

```python
det=PDE_Jacobian_3D.det()

det_expanded = sp.expand(det)
det_collected = sp.collect(det_expanded, x)
det_collected
```

## 4 Species System

```python
# Define symbolic matrix A
ODE_Jacobian_4D = sp.Matrix([
    [-A-x, -B, 0, E],
    [A, B-C-D-x, -G, 0],
    [0, D, 0-x, 0],
    [0, C, F, -E-x],
])

PDE_Jacobian_4D = sp.Matrix([
    [-A-d_1*(k**2)-x, -B, 0, E],
    [A, B-C-D-d_2*(k**2)-x, -G, 0],
    [0, D, 0-d_3*(k**2)-x, 0],
    [0, C, F, -E-d_4*(k**2)-x],
])
```

```python
det=ODE_Jacobian_4D.det()


det_expanded = sp.expand(det)
det_collected = sp.collect(det_expanded, x)
det_collected
```

```python
det=PDE_Jacobian_4D.det()


det_expanded = sp.expand(det)
det_collected = sp.collect(det_expanded, x)
det_collected
```
